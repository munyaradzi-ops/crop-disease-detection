# AI-Based Crop Disease Detection

## 02 - Dataset Preprocessing & Splitting

This notebook automates splitting the verified PlantVillage images into structured 
Training, Validation, and Test sets within the project's root `data/` folder.

### Splitting Strategy
- **Training (80%):** Feeds the neural network during training loops.
- **Validation (10%):** Monitored during training to adjust hyperparameters and catch overfitting early.
- **Testing (10%):** Kept completely unseen to evaluate real-world performance at the end.
- **Reproducibility:** Seeded with a random state to guarantee identical splits across runs.


```python
import os
from pathlib import Path
import splitfolders

# Define workspace roots using your verified configuration
PROJECT_ROOT = Path("..").resolve()
PLANTVILLAGE_DIR = PROJECT_ROOT.parent / "PlantVillage-Dataset"
SOURCE_DIR = PLANTVILLAGE_DIR / "raw" / "color"
OUTPUT_DIR = PROJECT_ROOT / "data"

print("Project root:        ", PROJECT_ROOT)
print("PlantVillage:        ", PLANTVILLAGE_DIR)
print("Source (Raw Data):   ", SOURCE_DIR)
print("Output (Split Data): ", OUTPUT_DIR)

# Verify source exists before running the heavy operation
assert SOURCE_DIR.exists(), f"❌ Cannot find source folder: {SOURCE_DIR}"
print("\n✅ Source path verified successfully. Ready to split folders.")
```

```python
print("⚡ Splitting dataset... This may take a moment depending on your storage drive.")

# splitfolders.ratio documentation parameters:
# - input: path to source directory containing subdirectories
# - output: target root folder where 'train', 'val', and 'test' will be written
# - seed: guarantees exact same distribution every time you run it
# - ratio: tuple mapping out percentages (Train, Val, Test)
splitfolders.ratio(
    input=str(SOURCE_DIR),
    output=str(OUTPUT_DIR),
    seed=1337,
    ratio=(0.8, 0.1, 0.1),
    group_prefix=None,
    move=False # Copies files instead of moving them, keeping your original raw data safe!
)

print("\n🎉 Dataset successfully split into Train (80%), Val (10%), and Test (10%)!")
```

```python
print("Checking your 'data/' output directory:")
print("=" * 40)
for folder in OUTPUT_DIR.iterdir():
    if folder.is_dir():
        # Count total images inside this specific split subset
        total_subset_images = sum(1 for file in folder.rglob('*') if file.is_file())
        print(f"📁 {folder.name:<10} -> Total files: {total_subset_images}")
```
